```python
import requests

url = "https://ebuu.fa.ap1.oraclecloud.com/hcmRestApi/resources/latest/recruitingCEJobRequisitions"

params = {
    "onlyData": "true",
    "expand": "requisitionList.secondaryLocations",
    "finder": 'findReqs;siteNumber=CX,keyword="automation",limit=25,offset=0,sortBy=POSTING_DATES_DESC'
}

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
}

response = requests.get(url, params=params, headers=headers)

print(response.status_code)

data = response.json()

jobs = data["items"][0].get("requisitionList", [])

for i,job in enumerate(jobs):
    print(i)
    print(job["Title"], "-", job.get("PrimaryLocation"))
```
